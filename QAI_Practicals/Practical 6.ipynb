Title : Implementing a 5 qubit Quantum Fourier transform

````python
from qiskit import QuantumRegister, ClassicalRegister
from qiskit import QuantumCircuit, execute,IBMQ
from qiskit.tools.monitor import job_monitor
from qiskit.circuit.library import QFT
import numpy as np

pi = np.pi


provider = IBMQ.get_provider(hub='ibm-q')

backend = provider.get_backend('ibmq_qasm_simulator')

q = QuantumRegister(5,'q')
c = ClassicalRegister(5,'c')

circuit = QuantumCircuit(q,c)

circuit.x(q[4])
circuit.x(q[2])
circuit.x(q[0])
circuit.append(QFT(num_qubits=5, approximation_degree=0, do_swaps=True, inverse=False, insert_barriers=False, name='qft'), q)
circuit.measure(q,c)
circuit.draw(output='mpl', filename='qft1.png')
print(circuit)

job = execute(circuit, backend, shots=1000)

job_monitor(job)

counts = job.result().get_counts()

print("\n QFT Output")
print("-------------")
print(counts)
input()

q = QuantumRegister(5,'q')
c = ClassicalRegister(5,'c')

circuit = QuantumCircuit(q,c)

circuit.x(q[4])
circuit.x(q[2])
circuit.x(q[0])
circuit.append(QFT(num_qubits=5, approximation_degree=0, do_swaps=True, inverse=False, insert_barriers=False, name='qft'), q)
circuit.measure(q,c)
circuit.draw(output='mpl',filename='qft2.png')

print(circuit)

job = execute(circuit, backend, shots=1000)

job_monitor(job)

counts = job.result().get_counts()

print("\n QFT with inverse QFT Output")
print("------------------------------")
print(counts)
input()
````

````python
# This practical is about **Quantum Fourier Transform (QFT) using Qiskit and IBM Quantum Simulator**.

# It demonstrates how the Quantum Fourier Transform converts a quantum state from the computational basis into the Fourier basis. QFT is an important quantum operation used in quantum algorithms such as Shor's factoring algorithm and Quantum Phase Estimation.

# I'll structure it in the same Jupyter Notebook format as your previous practicals, with Aim, Theory, code explanations, Applications, Result, Conclusion, and Viva Questions.

# # **Quantum Fourier Transform (QFT) Using Qiskit**

# [Computación Cuántica — Project-oriented workshop in Julia computing](https://images.openai.com/static-rsc-4/FokHi0QpbPN-1MO0GMb4vHYXGrcvlWAr8qcE_zqf5Ho6bzDJdleWDeadDuPeSncqMTfjC6O34LnyEEBFJI6PXoUhjo1ck2ZPbL1gz6YchuXgEzZvVisg4QB7eMXvSsagBUWmAd4Udtm_qdJCREDMH2GvzUtDdLtbUK1EPnxPOYw?purpose=inline)

# [juliaforeducation.github.io](https://juliaforeducation.github.io/Intro-Julia-2021/Computaci%C3%B3n_Cu%C3%A1ntica.html?utm_source=chatgpt.com)

# ## **1. Aim**

# To implement the Quantum Fourier Transform (QFT) on a 5-qubit quantum circuit using Qiskit, execute it on a quantum simulator, and observe the measurement outcomes.

# ## **2. Theory**

# The Quantum Fourier Transform is a quantum operation that transforms a quantum state from the computational basis into the Fourier basis.

# It is the quantum equivalent of the classical Discrete Fourier Transform (DFT), but it operates on quantum amplitudes.

# The QFT is mathematically represented as:

# \\[ QFT|x\rangle = \frac{1}{\sqrt{N}} \sum\_{y=0}^{N-1} e^{2\pi ixy/N}|y\rangle \\]

# Where:

# - \\(N=2^n\\), the number of possible states.
# - \\(n\\) is the number of qubits.
# - \\(x\\) is the input state.
# - \\(y\\) represents the output basis states.
# - \\(i\\) is the imaginary unit.

# In this practical, we use five qubits.

# Therefore:

# \\[ N=2^5=32 \\]

# The QFT transforms the input quantum state into a superposition of 32 possible computational basis states with specific amplitudes and phases.

# **Main objective:** To understand the implementation and working of QFT using quantum gates and observe its measurement distribution.

# ## **3. Import Required Libraries**

# Import the required Qiskit libraries to create quantum registers, construct quantum circuits, perform QFT operations, execute circuits, and visualize results.

# **Code Explanation:**

# - `QuantumRegister`: Creates quantum registers containing qubits.
# - `ClassicalRegister`: Creates classical registers to store measurement results.
# - `QuantumCircuit`: Used to create quantum circuits.
# - `execute`: Executes the quantum circuit on a simulator or backend.
# - `IBMQ`: Used to connect to IBM Quantum services in older Qiskit versions.
# - `job_monitor`: Monitors the execution status of a quantum job.
# - `QFT`: Implements the Quantum Fourier Transform.
# - `numpy`: Provides mathematical functions and constants.
# - `pi`: Represents the mathematical constant π.

# ## **4. Define the Mathematical Constant**

# ```
# pi = np.pi
# ```

# **Explanation:**

# - `np.pi` returns the mathematical value of π (approximately 3.14159).
# - The value is stored in the variable `pi`.
# - It can be used in quantum circuit calculations involving rotation angles and phase operations.

# **Note:** In this particular program, the variable `pi` is defined but not directly used.

# ## **5. Connect to IBM Quantum Provider**

# ```
# provider = IBMQ.get_provider(hub='ibm-q')
# ```

# **Explanation:**

# - `IBMQ.get_provider()` retrieves the IBM Quantum provider.
# - `hub='ibm-q'` specifies the IBM Quantum hub.
# - `provider` stores the connection information used to access available backends.

# The IBM Quantum provider allows users to access quantum computing resources through IBM's infrastructure.

# ## **6. Select the Quantum Simulator Backend**

# ```
# backend = provider.get_backend('ibmq_qasm_simulator')
# ```

# **Explanation:**

# - `provider.get_backend()` selects the required backend.
# - `'ibmq_qasm_simulator'` specifies the IBM Quantum QASM simulator.
# - The selected backend is stored in the `backend` variable.

# The simulator allows quantum circuits to be executed without requiring a physical quantum processor.

# ## **7. Create Quantum and Classical Registers**

# ```
# q = QuantumRegister(5, 'q')c = ClassicalRegister(5, 'c')
# ```

# **Explanation:**

# - `QuantumRegister(5, 'q')` creates a quantum register containing five qubits.
# - `ClassicalRegister(5, 'c')` creates a classical register containing five bits.
# - `q` represents the quantum register.
# - `c` represents the classical register.

# Since five qubits are used, the total number of possible computational basis states is:

# \\[ 2^5=32 \\]

# ## **8. Create a Quantum Circuit**

# ```
# circuit = QuantumCircuit(q, c)
# ```

# **Explanation:**

# - `QuantumCircuit()` creates a quantum circuit.
# - `q` specifies the five quantum bits.
# - `c` specifies the five classical bits.
# - The circuit will contain quantum operations and measurement instructions.

# ## **9. Prepare the Initial Quantum State**

# ```
# circuit.x(q[4])circuit.x(q[2])circuit.x(q[0])
# ```

# **Explanation:**

# These statements apply Pauli-X gates to selected qubits.

# - `circuit.x(q[4])` flips qubit 4 from 0 to 1.
# - `circuit.x(q[2])` flips qubit 2 from 0 to 1.
# - `circuit.x(q[0])` flips qubit 0 from 0 to 1.

# The Pauli-X gate works like a classical NOT gate.

# \\[ X|0\rangle=|1\rangle \\]

# \\[ X|1\rangle=|0\rangle \\]

# Initially, all qubits are in the state 0.

# After applying these three gates, the quantum register becomes:

# ```
# Qubit:  q4 q3 q2 q1 q0
# State:   1  0  1  0  1
# ```

# Therefore, the initial state is:

# \\[ |10101\rangle \\]

# Its decimal equivalent is 21.

# ## **10. Apply the Quantum Fourier Transform**

# ```
# circuit.append(    QFT(        num_qubits=5,        approximation_degree=0,        do_swaps=True,        inverse=False,        insert_barriers=False,        name='qft'    ), q)
# ```

# **Explanation:**

# This is the most important operation in the practical.

# - `circuit.append()` adds a quantum operation to the existing circuit.
# - `QFT()` creates the Quantum Fourier Transform operation.
# - `num_qubits=5` specifies that QFT operates on five qubits.
# - `approximation_degree=0` uses the full QFT without omitting small-angle controlled-phase rotations.
# - `do_swaps=True` includes the final swap operations to arrange the output qubits in standard order.
# - `inverse=False` specifies that the forward QFT is applied.
# - `insert_barriers=False` means no additional barriers are inserted between QFT operations.
# - `name='qft'` assigns the operation the name `qft`.
# - `q` specifies the quantum register on which QFT is applied.

# The QFT converts the input state into a superposition of 32 computational basis states.

# The mathematical representation is:

# \\[ QFT|x\rangle = \frac{1}{\sqrt{32}} \sum\_{y=0}^{31} e^{2\pi ixy/32}|y\rangle \\]

# ## **11. Measure the Quantum Circuit**

# ```
# circuit.measure(q, c)
# ```

# **Explanation:**

# - `measure()` performs measurement on the quantum register.
# - `q` specifies all five qubits.
# - `c` specifies the five classical bits.
# - The measurement results are stored in the classical register.

# After measurement, the quantum state collapses into one of the 32 possible computational basis states.

# ## **12. Draw and Visualize the Circuit**

# ```
# circuit.draw(output='mpl', filename='qft1.png')print(circuit)
# ```

# **Explanation:**

# - `circuit.draw()` generates a graphical representation of the quantum circuit.
# - `output='mpl'` uses Matplotlib to draw the circuit.
# - `filename='qft1.png'` saves the circuit diagram as an image.
# - `print(circuit)` displays the circuit in text format.

# The circuit diagram helps visualize the initial X gates, QFT operations, and measurement gates.

# ## **13. Execute the Quantum Circuit**

# ```
# job = execute(circuit, backend, shots=1000)
# ```

# **Explanation:**

# - `execute()` sends the quantum circuit to the selected backend.
# - `circuit` specifies the quantum circuit to execute.
# - `backend` specifies the IBM QASM simulator.
# - `shots=1000` means the circuit is executed 1000 times.

# Executing multiple shots helps obtain the statistical distribution of measurement outcomes.

# ## **14. Monitor the Quantum Job**

# ```
# job_monitor(job)
# ```

# **Explanation:**

# - `job_monitor()` monitors the execution status of the quantum job.
# - `job` represents the submitted quantum circuit execution.
# - It displays the progress and status of the job.

# This is useful when executing circuits on remote quantum computing services.

# ## **15. Retrieve the Measurement Results**

# ```
# counts = job.result().get_counts()
# ```

# **Explanation:**

# - `job.result()` retrieves the execution results.
# - `get_counts()` extracts the measurement outcomes.
# - `counts` stores the outcomes in a Python dictionary.

# Example output:

# ```
# {    '00000': 31,    '00001': 34,    '00010': 29,    '00011': 32}
# ```

# The actual output can contain all 32 possible outcomes, with frequencies that vary between executions.

# ## **16. Display the QFT Output**

# ```
# print("\n QFT Output")print("-------------")print(counts)input()
# ```

# **Explanation:**

# - `print("\n QFT Output")` displays the heading.
# - `print("-------------")` displays a separator.
# - `print(counts)` displays the measurement results.
# - `input()` pauses program execution until the user presses Enter.

# The output represents the measurement distribution obtained after applying the QFT.

# ## **17. Create the Second Quantum Circuit**

# The program creates another quantum circuit using the same five-qubit and five-classical-bit registers.

# ```
# q = QuantumRegister(5, 'q')c = ClassicalRegister(5, 'c')circuit = QuantumCircuit(q, c)
# ```

# **Explanation:**

# - Creates a new quantum register containing five qubits.
# - Creates a new classical register containing five bits.
# - Initializes a new quantum circuit using these registers.

# This second circuit is intended to demonstrate another QFT-related operation.

# ## **18. Prepare the Initial State Again**

# ```
# circuit.x(q[4])circuit.x(q[2])circuit.x(q[0])
# ```

# **Explanation:**

# The same Pauli-X gates are applied to qubits 4, 2, and 0.

# This prepares the same initial quantum state:

# \\[ |10101\rangle \\]

# ## **19. Apply QFT Again**

# ```
# circuit.append(    QFT(        num_qubits=5,        approximation_degree=0,        do_swaps=True,        inverse=False,        insert_barriers=False,        name='qft'    ), q)
# ```

# **Explanation:**

# The same forward Quantum Fourier Transform is applied to the second quantum circuit.

# The parameter `inverse=False` indicates that this is a forward QFT operation.

# **Important correction:** Your second circuit is identical to the first circuit. Despite the output heading saying "QFT with inverse QFT Output", it does not actually perform an inverse QFT.

# To perform an inverse QFT, change:

# ```
# inverse=False
# ```

# to:

# ```
# inverse=True
# ```

# ## **20. Measure and Visualize the Second Circuit**

# ```
# circuit.measure(q, c)circuit.draw(output='mpl', filename='qft2.png')print(circuit)
# ```

# **Explanation:**

# - `circuit.measure(q, c)` measures all five qubits.
# - `circuit.draw()` generates the second circuit diagram.
# - `filename='qft2.png'` saves the diagram.
# - `print(circuit)` displays the circuit structure.

# ## **21. Execute the Second Circuit**

# ```
# job = execute(circuit, backend, shots=1000)job_monitor(job)counts = job.result().get_counts()
# ```

# **Explanation:**

# - Executes the second circuit using the selected backend.
# - Uses 1000 shots to collect measurement outcomes.
# - `job_monitor()` displays the execution status.
# - `get_counts()` retrieves the measurement statistics.

# ## **22. Display the Second Output**

# ```
# print("\n QFT with inverse QFT Output")print("------------------------------")print(counts)input()
# ```

# **Explanation:**

# - Displays the heading for the second experiment.
# - Prints the measurement counts.
# - `input()` pauses the execution until Enter is pressed.

# **Note:** With your current code, this is another forward QFT output, not an inverse QFT result.

# ## **23. Applications of Quantum Fourier Transform**

# **1. Shor's Factoring Algorithm**

# QFT is an important component of Shor's algorithm, which is designed for integer factorization.

# **2. Quantum Phase Estimation**

# QFT is used to extract phase information from quantum states and is a key component of quantum phase estimation.

# **3. Quantum Signal Processing**

# QFT can be used in quantum algorithms involving signal and frequency transformations.

# **4. Quantum Chemistry**

# It plays a role in quantum algorithms used for estimating eigenvalues and studying molecular systems.

# **5. Period Finding**

# QFT helps identify periodic patterns in quantum algorithms.

# ## **24. Result**

# The Quantum Fourier Transform was implemented on a five-qubit quantum circuit using Qiskit.

# The initial quantum state was prepared using Pauli-X gates, followed by the application of QFT. The circuit was executed for 1000 shots, and the measurement outcomes were collected and displayed.

# The experiment demonstrates how QFT transforms a computational basis state into a superposition of quantum states.

# The second circuit in the supplied code also applies forward QFT. An inverse QFT requires changing the `inverse` parameter to `True`.

# ## **25. Conclusion**

# This practical demonstrates the implementation of the Quantum Fourier Transform using Qiskit.

# It introduces quantum registers, classical registers, Pauli-X gates, QFT operations, quantum measurement, and simulator execution.

# QFT is an important building block of several quantum algorithms, particularly Shor's factoring algorithm and Quantum Phase Estimation.

# ## **26. Oral Questions (Viva Questions and Answers)**

# **Q1. What is QFT?**

# QFT stands for Quantum Fourier Transform. It is a quantum operation that transforms a quantum state from the computational basis into the Fourier basis.

# **Q2. What is the aim of this practical?**

# To implement the Quantum Fourier Transform on a five-qubit quantum circuit using Qiskit and observe its measurement outcomes.

# **Q3. What is the classical equivalent of QFT?**

# The classical equivalent is the Discrete Fourier Transform (DFT).

# **Q4. How many possible states can five qubits represent?**

# Five qubits can represent:

# \\[ 2^5=32 \\]

# computational basis states.

# **Q5. What is the mathematical formula of QFT?**

# \\[ QFT|x\rangle = \frac{1}{\sqrt{N}} \sum\_{y=0}^{N-1} e^{2\pi ixy/N}|y\rangle \\]

# **Q6. What is the purpose of QFT?**

# QFT transforms quantum amplitudes and phases into the Fourier basis. It is useful for extracting periodicity and phase information in quantum algorithms.

# **Q7. What is a QuantumRegister?**

# A QuantumRegister is a collection of qubits used in a quantum circuit.

# **Q8. What is a ClassicalRegister?**

# A ClassicalRegister is a collection of classical bits used to store measurement outcomes.

# **Q9. What is a Pauli-X gate?**

# A Pauli-X gate flips a qubit from 0 to 1 or from 1 to 0. It is equivalent to a classical NOT gate on computational basis states.

# **Q10. Why are X gates used in this practical?**

# They prepare the initial quantum state by changing selected qubits from 0 to 1.

# **Q11. What is the initial state in this practical?**

# The initial state is:

# \\[ |10101\rangle \\]

# **Q12. What does do_swaps=True mean?**

# It includes the final swap operations in the QFT circuit to arrange the output qubits in standard order.

# **Q13. What does inverse=False mean?**

# It specifies that the forward Quantum Fourier Transform is applied.

# **Q14. What is an inverse QFT?**

# An inverse QFT reverses the operation of the forward QFT and transforms a Fourier-basis state back into its original computational-basis state.

# **Q15. What is the purpose of shots=1000?**

# It executes the quantum circuit 1000 times to obtain measurement statistics.

# **Q16. What does get_counts() do?**

# It returns a dictionary containing the measured binary outcomes and their frequencies.

# **Q17. What is the purpose of IBM Quantum?**

# IBM Quantum provides access to quantum computing tools, simulators, and quantum hardware through its cloud platform.

# **Q18. What is a quantum simulator?**

# A quantum simulator is software that imitates quantum circuit operations using classical computing resources.

# **Q19. What is a Fourier basis?**

# The Fourier basis is a set of quantum states with specific phase relationships produced by the Quantum Fourier Transform.

# **Q20. What is the relationship between QFT and Shor's algorithm?**

# QFT is used in Shor's algorithm to extract periodicity information, which helps in integer factorization.

# **Q21. What is the relationship between QFT and Quantum Phase Estimation?**

# QFT, particularly inverse QFT, is used to extract phase information from quantum states in Quantum Phase Estimation.

# **Q22. What is the difference between QFT and inverse QFT?**

# QFT transforms computational basis states into Fourier basis states, while inverse QFT performs the reverse transformation.

# **Q23. What is the purpose of transpilation?**

# Transpilation converts and optimizes quantum circuits for a selected quantum backend.

# **Q24. What happens if we apply QFT followed by inverse QFT?**

# The original quantum state is recovered in an ideal quantum circuit.

# \\[ QFT^\dagger QFT=I \\]

# **Q25. What is the final output of this practical?**

# The program displays the measurement counts obtained after applying QFT to a five-qubit quantum circuit.

# ## **27. Important Formulas**

# **1. Number of possible quantum states**

# \\[ N=2^n \\]

# For five qubits:

# \\[ N=2^5=32 \\]

# **2. Quantum Fourier Transform**

# \\[ QFT|x\rangle = \frac{1}{\sqrt N} \sum\_{y=0}^{N-1} e^{2\pi ixy/N}|y\rangle \\]

# **3. Inverse Quantum Fourier Transform**

# \\[ QFT^{-1}|y\rangle = \frac{1}{\sqrt N} \sum\_{x=0}^{N-1} e^{-2\pi ixy/N}|x\rangle \\]

# **4. QFT and inverse QFT relationship**

# \\[ QFT^{-1}\times QFT=I \\]

# ## **28. Quick Revision Table**

# | Concept           | Meaning                               |
# | ----------------- | ------------------------------------- |
# | QFT               | Quantum Fourier Transform             |
# | DFT               | Discrete Fourier Transform            |
# | QFT inverse       | Reverse Fourier transformation        |
# | QuantumRegister   | Stores qubits                         |
# | ClassicalRegister | Stores measurement results            |
# | Pauli-X           | Bit-flip gate                         |
# | IBMQ              | Legacy IBM Quantum provider interface |
# | QASM Simulator    | Quantum circuit simulator             |
# | Shots             | Number of circuit executions          |
# | do_swaps          | Includes final swap gates             |
# | Transpilation     | Circuit optimization                  |

# ## **Important corrections before running your practical**

# There are three things to remember:

# 1. **Both circuits are currently identical.** Your second circuit uses `inverse=False`, so it performs forward QFT again.
# 2. **The output heading is misleading.** To demonstrate inverse QFT properly, use `inverse=True` in the second circuit, or better, apply forward QFT followed by inverse QFT in the same circuit before measurement.
# 3. **IBM Qiskit compatibility:** `IBMQ` and `execute()` belong to older Qiskit APIs. Modern Qiskit uses updated IBM Quantum account and backend interfaces. If you encounter import errors, the code will need updating.

# **Expected observation:** Since the initial state is a computational basis state, ideal QFT produces equal measurement probabilities across all 32 outcomes. With 1000 shots, each outcome should appear approximately 31–32 times, although individual counts will vary.

# For your viva, particularly understand the difference between forward QFT and inverse QFT, why five qubits produce 32 states, and how QFT is used in Shor's algorithm.
````
