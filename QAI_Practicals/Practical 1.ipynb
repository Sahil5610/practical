Title : Implementation of 16 Qubit Random Number Generator

```python
#16qubit random number genertor
from qiskit import QuantumCircuit, Aer, execute
from qiskit.visualization import circuit_drawer

# Create a quantum circuit with 16 qubits
circuit = QuantumCircuit(16, 16)

# Apply Hadamard gates to put all qubits in superposition
circuit.h(range(16))

# Measure all qubits
circuit.measure(range(16), range(16))

# Visualize the circuit
print(circuit)
circuit_drawer(circuit, output='mpl')

# Simulate the quantum circuit using the QASM simulator
simulator = Aer.get_backend('qasm_simulator')
job = execute(circuit, simulator, shots=1)
result = job.result()
counts = result.get_counts(circuit)

# Extract the random number from the measurement outcome
random_number = int(list(counts.keys())[0], 2)

# Convert the random number to binary representation
binary_number = bin(random_number)[2:].zfill(16)

print("Random number (decimal):", random_number)
print("Random number (binary):", binary_number)
```

```python
# # 16-Qubit Quantum Random Number Generator

# ### 1. Import Required Libraries

# Import the necessary Qiskit libraries to create quantum circuits, execute quantum operations, simulate the circuit, and visualize the circuit diagram.

# - `QuantumCircuit`: Used to create a quantum circuit.
# - `Aer`: Provides quantum simulation backends.
# - `execute`: Used to execute the quantum circuit.
# - `circuit_drawer`: Used to visualize the quantum circuit.

# ### 2. Create a Quantum Circuit

# Create a quantum circuit containing 16 qubits and 16 classical bits. Qubits are used for quantum computation, while classical bits store the measurement results.

# **Code Explanation:**
# - `QuantumCircuit(16, 16)` creates a circuit with 16 quantum bits and 16 classical bits.
# - The first argument represents the number of qubits.
# - The second argument represents the number of classical bits.

# ### 3. Apply Hadamard Gates

# Apply Hadamard (H) gates to all 16 qubits using `circuit.h(range(16))`. The Hadamard gate puts each qubit into an equal superposition of 0 and 1, giving both outcomes a 50% measurement probability.

# **Code Explanation:**
# - `circuit.h()` applies the Hadamard gate.
# - `range(16)` selects all 16 qubits.
# - Each qubit has a 50% probability of measuring 0 or 1.

# Therefore, the circuit represents:

# 2^16 = 65,536 possible measurement outcomes.

# ### 4. Measure All Qubits

# Measure all 16 qubits and store their results in the corresponding classical bits. Measurement collapses the quantum superposition into one definite 16-bit binary outcome.

# **Code Explanation:**
# - `circuit.measure()` performs the measurement operation.
# - `range(16)` selects all 16 qubits.
# - The second `range(16)` specifies the corresponding classical bits where the measurement results are stored.

# ### 5. Visualize the Quantum Circuit

# Display the quantum circuit using `print(circuit)` and `circuit_drawer()`. This helps visualize the Hadamard gates and measurement operations applied to each qubit.

# **Code Explanation:**
# - `print(circuit)` displays the circuit in text format.
# - `circuit_drawer(circuit, output='mpl')` generates a graphical circuit representation using Matplotlib.
# - The output helps understand the arrangement of quantum gates and measurements.

# ### 6. Initialize the Quantum Simulator

# Select the QASM simulator from Qiskit Aer. The simulator executes the quantum circuit on a classical computer without requiring access to physical quantum hardware.

# **Code Explanation:**
# - `Aer.get_backend()` selects the required simulation backend.
# - `'qasm_simulator'` specifies the quantum assembly simulator.
# - The simulator is stored in the `simulator` variable.

# ### 7. Execute the Quantum Circuit

# Execute the circuit using `shots=1`. One shot means the circuit is executed once, producing one 16-bit measurement result.

# **Code Explanation:**
# - `execute()` sends the quantum circuit to the simulator.
# - `circuit` specifies the circuit to execute.
# - `simulator` specifies the execution backend.
# - `shots=1` means the circuit is executed only once.

# ### 8. Retrieve Measurement Results

# Retrieve the execution results using `job.result()` and extract the measurement counts using `get_counts()`. The output contains the measured binary string and its frequency.

# **Code Explanation:**
# - `job.result()` retrieves the execution results.
# - `result.get_counts(circuit)` returns the measurement outcomes and their frequencies.
# - The output is stored in the `counts` dictionary.

# Example output:

# `{'1011010010110101': 1}`

# Here, the binary outcome was measured once.

# ### 9. Convert Binary into Decimal

# Extract the measured binary string from the counts dictionary and convert it into a decimal integer using Python's `int()` function with base 2.

# **Code Explanation:**
# - `counts.keys()` extracts the measurement outcome.
# - `list(counts.keys())[0]` selects the first binary result.
# - `int(binary_string, 2)` converts the binary string into a decimal number.
# - The generated decimal number is stored in `random_number`.

# ### 10. Convert Decimal into 16-Bit Binary

# Convert the generated decimal number back into binary using `bin()`. Remove the `0b` prefix using `[2:]` and use `zfill(16)` to ensure the binary representation contains exactly 16 bits.

# **Code Explanation:**
# - `bin(random_number)` converts the decimal number into binary.
# - `[2:]` removes the `0b` prefix.
# - `zfill(16)` adds leading zeros if required to make the binary string exactly 16 bits long.
# - The final binary number is stored in `binary_number`.

# ### 11. Display the Final Output

# Print the generated random number in both decimal and binary formats.

# **Code Explanation:**
# - `print()` displays the generated decimal number.
# - The second `print()` displays the corresponding 16-bit binary representation.

# ### Result

# The program generates a 16-bit random number using quantum superposition and measurement. Since 16 qubits can represent 2^16 = 65,536 possible measurement outcomes, the generated decimal number ranges from 0 to 65,535.

# **Note:** This implementation uses a quantum simulator rather than actual quantum hardware. The simulator samples the quantum measurement probabilities using classical computation.
```
